```python
words = ["a", "b", "a", "c", "a"]

# result -> {a:3, b:1, c:1}

counts = {}

for w in words:
    counts[w] = counts.get(w, 0) + 1

print(counts)
```

```python
from collections import Counter

words = ["a", "b", "a", "c", "a"]

print(Counter(words))
```

```python
conditions = {
    "road1": "good",
    "road2": "poor",
    "road3": "good",
    "road4": "moderate",
    "road5": "poor"
}

print(conditions)

ratings = []
for _ , val in conditions.items():
    ratings.append(val)

print(ratings)

from collections import Counter

rating_dict = Counter(ratings)


roads = []
road_ratings = []

for key, val in rating_dict.items():
    roads.append(key)
    road_ratings.append(val)

print(roads, road_ratings)
```

```python
import pandas as pd
```

```python
pd.DataFrame({"Road": roads, "Ratings": road_ratings})
```
